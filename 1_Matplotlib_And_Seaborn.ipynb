```python
from matplotlib.lines import lineStyles
#Sub Plot
x=[1,2,3,4,5,6]
y1=[9,8,7,4,5,6]
y2=[4,5,6,3,1,8]
plt.figure(figsize=(6,4))
plt.subplot(2,2,1)
plt.plot(x,y1,color=("pink"),marker=("*"),markersize=("10"))
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Sell Oder Plot")
plt.grid(True)

plt.subplot(2,2,2,)
plt.plot(x,y2,color=("black"),marker=("o"),markersize=("10"))
plt.grid(True)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Production Plot")
plt.show()
```

```python
high=df.groupby("top_scorer")["highscore"].sum().sort_values(ascending=False).head(3)
high
```

```python
high.plot(kind="bar")
```

```python
df["highest_wickets"]=df["best_bowling_figure"].apply(lambda x:x.split("--")[0])
df["highest_wickets"]=df["highest_wickets"].astype(int)
top_bowlers=df.groupby("best_bowling")["highest_wickets"].sum().sort_values(ascending=False).head(5)
top_bowlers.plot(kind="bar")
```
